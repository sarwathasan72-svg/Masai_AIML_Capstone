```python
from sklearn.datasets import load_iris
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

# Load data
data = load_iris()
X = data.data
y = data.target

# Same folds for both models
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Logistic Regression
lg_model = LogisticRegression(max_iter=2000)

lg_scores = cross_val_score(
    lg_model, X, y, cv=cv, scoring="accuracy"
)

# Decision Tree
dt_model = DecisionTreeClassifier(random_state=42)

dt_scores = cross_val_score(
    dt_model, X, y, cv=cv, scoring="accuracy"
)

# Compare results
print("Logistic Regression fold scores:", lg_scores.round(3))
print("Logistic Regression mean accuracy:", round(lg_scores.mean(), 3))
print("Logistic Regression std:", round(lg_scores.std(), 3))

print("\nDecision Tree fold scores:", dt_scores.round(3))
print("Decision Tree mean accuracy:", round(dt_scores.mean(), 3))
print("Decision Tree std:", round(dt_scores.std(), 3))

if lg_scores.mean() > dt_scores.mean():
    print("\nLogistic Regression has better average CV accuracy.")
elif dt_scores.mean() > lg_scores.mean():
    print("\nDecision Tree has better average CV accuracy.")
else:
    print("\nBoth models have the same average CV accuracy.")
```

```python
import numpy as np

def boost_round(X, y, pred, eta, threshold):

    # Step 1: Convert Python lists into NumPy arrays
    # NumPy lets us subtract arrays and select values using conditions.
    X = np.array(X)
    y = np.array(y)
    pred = np.array(pred, dtype=float)

    # Step 2: Calculate residuals (errors)
    # Residual = actual score - predicted score
    residuals = y - pred

    print("Step 2: Residuals")
    print(residuals)
    # [-24, -14, -4, 16, 26]

    # Step 3: Identify which students belong to each group
    left_group = X <= threshold
    right_group = X > threshold

    print("\nStep 3: Group selection")
    print("Left group :", left_group)
    print("Right group:", right_group)
    # Left:  [True, True, True, False, False]
    # Right: [False, False, False, True, True]

    # Step 4: Select the residuals belonging to each group
    left_residuals = residuals[left_group]
    right_residuals = residuals[right_group]

    print("\nStep 4: Residuals in each group")
    print("Left :", left_residuals)
    print("Right:", right_residuals)
    # Left:  [-24, -14, -4]
    # Right: [16, 26]

    # Step 5: Calculate each group's average residual
    left_average = left_residuals.mean()
    right_average = right_residuals.mean()

    print("\nStep 5: Average residuals")
    print("Left :", left_average)       # -14.0
    print("Right:", right_average)      # 21.0

    # Step 6: Multiply each average by the learning rate
    left_correction = eta * left_average
    right_correction = eta * right_average

    print("\nStep 6: Corrections")
    print("Left :", left_correction)    # -7.0
    print("Right:", right_correction)   # 10.5

    # Step 7: Select the current predictions for each group
    left_predictions = pred[left_group]
    right_predictions = pred[right_group]

    print("\nStep 7: Current predictions")
    print("Left :", left_predictions)   # [64, 64, 64]
    print("Right:", right_predictions)  # [64, 64]

    # Step 8: Add the corrections to the current predictions
    new_left_predictions = left_predictions + left_correction
    new_right_predictions = right_predictions + right_correction

    print("\nStep 8: Updated group predictions")
    print("Left :", new_left_predictions)   # [57, 57, 57]
    print("Right:", new_right_predictions)  # [74.5, 74.5]

    # Step 9: Copy the original predictions
    updated_predictions = pred.copy()

    # Step 10: Place updated values into the correct positions
    updated_predictions[left_group] = new_left_predictions
    updated_predictions[right_group] = new_right_predictions

    print("\nStep 10: All updated predictions")
    print(updated_predictions)
    # [57.0, 57.0, 57.0, 74.5, 74.5]

    # Step 11: Convert the NumPy array into a Python list
    result = updated_predictions.tolist()

    # Step 12: Return the final list
    return result


# Input data
X = [1, 2, 3, 4, 5]          # Study hours
y = [40, 50, 60, 80, 90]     # Actual exam scores
pred = [64, 64, 64, 64, 64]  # Current predictions

eta = 0.5                    # Apply half the correction
threshold = 3                # Split at 3 study hours

# Perform one boosting round
new_pred = boost_round(X, y, pred, eta, threshold)

print("\nFinal result:", new_pred)
```

```python
import math

POINTS = [(2, 3), (3, 3), (8, 8), (9, 9)]
CENTROIDS = [(2, 2), (9, 8)]


# Find the distance between one point and one centroid
def euclidean_distance(point, centroid):
    x_difference = point[0] - centroid[0]
    y_difference = point[1] - centroid[1]

    distance = math.sqrt(
        x_difference ** 2 + y_difference ** 2
    )

    return distance


# Assign each point to cluster 0 or cluster 1
# This simple version works with exactly two centroids.
def assign_clusters(points, centroids):
    labels = []

    for point in points:
        distance_to_first = euclidean_distance(point, centroids[0])

        # point(" 1: ",point.tolist)
        distance_to_second = euclidean_distance(point, centroids[1])

        # point(" 2: ",point.tolist)

        if distance_to_first <= distance_to_second:
            labels.append(0)
        else:
            labels.append(1)

    return labels


# Calculate the average position for each cluster
def update_centroids(points, labels, k):
    # Make space to store totals for each cluster
    x_totals = [0] * k
    y_totals = [0] * k
    counts = [0] * k

    # Add each point's coordinates to its cluster's totals
    for i in range(len(points)):
        point = points[i]
        cluster = labels[i]

        x_totals[cluster] += point[0]
        y_totals[cluster] += point[1]
        counts[cluster] += 1

    new_centroids = []

    # Divide each cluster's totals by its number of points
    for cluster in range(k):
        if counts[cluster] == 0:
            raise ValueError("A cluster has no points to average.")

        average_x = x_totals[cluster] / counts[cluster]
        average_y = y_totals[cluster] / counts[cluster]

        new_centroids.append((average_x, average_y))

    return new_centroids


labels = assign_clusters(POINTS, CENTROIDS)
print("Labels:", labels)

new_centroids = update_centroids(POINTS, labels, k=2)
print("New centroids:", new_centroids)
```

```python
"""Synthetic shipment data for the Delay Intelligence capstone.
Run once:  python generate_data.py   (writes data/hubs.csv, data/raw_shipments.csv, data/new_shipments.csv)
Do not edit any constant: every number in the brief was measured on this exact output."""
import os
import numpy as np
import pandas as pd

HUBS = [
    ("HUB01", "Mumbai", "West", 1, 1500), ("HUB02", "Delhi", "North", 1, 1600),
    ("HUB03", "Bengaluru", "South", 1, 1450), ("HUB04", "Hyderabad", "South", 1, 1350),
    ("HUB05", "Chennai", "South", 1, 1300), ("HUB06", "Pune", "West", 1, 1250),
    ("HUB07", "Kolkata", "East", 1, 1200), ("HUB08", "Ahmedabad", "West", 1, 1150),
    ("HUB09", "Jaipur", "North", 2, 950), ("HUB10", "Lucknow", "North", 2, 900),
    ("HUB11", "Chandigarh", "North", 2, 850), ("HUB12", "Indore", "West", 2, 820),
    ("HUB13", "Nagpur", "West", 2, 780), ("HUB14", "Coimbatore", "South", 2, 740),
    ("HUB15", "Kochi", "South", 2, 700), ("HUB16", "Surat", "West", 2, 650),
    ("HUB17", "Patna", "East", 3, 520), ("HUB18", "Ranchi", "East", 3, 480),
    ("HUB19", "Guwahati", "East", 3, 450), ("HUB20", "Raipur", "East", 3, 420),
    ("HUB21", "Bhubaneswar", "East", 3, 390), ("HUB22", "Dehradun", "North", 3, 350),
    ("HUB23", "Siliguri", "East", 3, 310), ("HUB24", "Agartala", "East", 3, 270),
]
DIST_MU = [180, 320, 480]      # mean lane distance (km) for hub tier 1 / 2 / 3
DIST_SD = [50, 90, 150]
STOPS_MU = [55, 70, 85]        # mean stops on the rider route for hub tier 1 / 2 / 3
HUB_EFFECT = [0, 2, 4]         # extra hours for hub tier 1 / 2 / 3
# How a hub really operates (1 = metro-like, 2 = regional, 3 = stretched). Four hubs operate unlike their tier label.
OPERATING_TIER = {"HUB07": 2, "HUB13": 3, "HUB15": 3, "HUB22": 2}
BASE_HOURS = {"Express": 10, "Standard": 28, "Economy": 58}


def simulate(n_per_hub, seed):
    """Clean shipments, n_per_hub rows for each of the 24 hubs, in hub order."""
    np.random.seed(seed)
    hubs = pd.DataFrame(HUBS, columns=["hub_id", "hub_city", "region", "hub_tier", "hub_daily_capacity"])
    n = n_per_hub * len(hubs)
    op_tier = [OPERATING_TIER.get(h, t) for h, t in zip(hubs["hub_id"], hubs["hub_tier"])]
    tier = np.repeat(op_tier, n_per_hub) - 1
    hub_id = np.repeat(hubs["hub_id"].values, n_per_hub)
    distance = np.clip(np.random.normal(np.array(DIST_MU)[tier], np.array(DIST_SD)[tier]), 15, 1500).round(1)
    stops = np.clip(np.random.normal(np.array(STOPS_MU)[tier], 10), 20, 140).round().astype(int)
    weight = np.clip(np.random.lognormal(0.6, 0.7, n), 0.1, 40).round(2)
    value = np.random.lognormal(7.0, 0.8, n).round().astype(int)
    service = np.random.choice(["Express", "Standard", "Economy"], n, p=[0.2, 0.55, 0.25])
    pay = np.random.choice(["Prepaid", "COD"], n, p=[0.6, 0.4])
    rain = (np.random.rand(n) < 0.18).astype(int)
    day = np.random.randint(0, 90, n)
    hour = np.random.randint(6, 22, n)
    minute = np.random.randint(0, 60, n)
    ts = pd.Timestamp("2026-07-01") + pd.to_timedelta(day, unit="D") + pd.to_timedelta(hour * 60 + minute, unit="m")
    sale = ((ts >= pd.Timestamp("2026-09-15")) & (ts < pd.Timestamp("2026-09-29"))).astype(int)
    weekend = (ts.dayofweek >= 5).astype(int)
    hours = (np.array([BASE_HOURS[s] for s in service]) + 0.03 * distance + 0.10 * stops
             + np.array(HUB_EFFECT)[tier] + 0.12 * weight + 4 * rain + 7 * sale
             + 3 * (pay == "COD") + 2 * weekend + 3 * (hour >= 18)
             + 12 * sale * (service == "Express") + 8 * (distance > 400) * rain
             + np.random.normal(0, 4, n)).round(2)
    return pd.DataFrame({
        "hub_id": hub_id, "service_type": service, "payment_mode": pay,
        "parcel_weight_kg": weight, "distance_km": distance, "declared_value": value,
        "n_stops": stops, "rain_flag": rain, "sale_event_flag": sale, "dispatch_time": ts,
        "label_print_batch": np.random.randint(1000, 9999, n),
        "scanner_build_no": np.random.randint(1, 40, n),
        "tape_roll_no": np.random.randint(1, 500, n),
        "sticker_serial_mod": np.random.uniform(0, 1, n).round(3),
        "shelf_slot_no": np.random.randint(1, 300, n),
        "delivery_hours": hours,
    })


def make_messy(df, seed, n_missing, n_case, n_sentinel, n_dupes, id_start):
    """Turn a clean table into a raw export: text columns, gaps, duplicates, a sentinel."""
    np.random.seed(seed)
    df = df.sample(frac=1.0, random_state=seed).reset_index(drop=True)
    n = len(df)
    df.insert(0, "shipment_id", ["SHP" + str(id_start + i) for i in range(n)])
    df["dispatch_time"] = df["dispatch_time"].dt.strftime("%Y-%m-%d %H:%M")
    df["declared_value"] = df["declared_value"].map(lambda v: "\u20b9" + format(v, ","))
    for col, k in [("distance_km", n_missing[0]), ("parcel_weight_kg", n_missing[1]),
                   ("declared_value", n_missing[2]), ("payment_mode", n_missing[3])]:
        df.loc[np.random.choice(n, k, replace=False), col] = np.nan
    pick = np.random.choice(n, n_case, replace=False)
    shapes = np.random.randint(0, 3, n_case)
    for i, s in zip(pick, shapes):
        v = df.at[i, "service_type"]
        df.at[i, "service_type"] = [v.upper(), v.lower(), "  " + v + " "][s]
    if n_sentinel:
        df.loc[np.random.choice(n, n_sentinel, replace=False), "parcel_weight_kg"] = 999.0
    if n_dupes:
        df = pd.concat([df, df.iloc[np.random.choice(n, n_dupes, replace=False)]], ignore_index=True)
    return df


if __name__ == "__main__":
    os.makedirs("data", exist_ok=True)
    hubs = pd.DataFrame(HUBS, columns=["hub_id", "hub_city", "region", "hub_tier", "hub_daily_capacity"])
    hubs.to_csv("data/hubs.csv", index=False)
    raw = make_messy(simulate(50, 2604), seed=7, n_missing=(36, 24, 12, 18),
                     n_case=120, n_sentinel=5, n_dupes=24, id_start=10001)
    raw.to_csv("data/raw_shipments.csv", index=False)
    new = make_messy(simulate(2, 99), seed=8, n_missing=(3, 2, 1, 2),
                     n_case=6, n_sentinel=0, n_dupes=0, id_start=90001)
    new = new.drop(columns=["delivery_hours"])
    new.to_csv("data/new_shipments.csv", index=False)
    print(len(hubs), len(raw), len(new))
```
